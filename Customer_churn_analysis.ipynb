```python
import pandas as pd
```

```python
df= pd.read_excel("customer_churn.xlsx")
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
## Find the number of churned customers
df["Churn"].value_counts()
```

```python
## Find the churn percentage
df["Churn"].value_counts(normalize=True)*100
```

```python
## To find the churn rate
churn_rate=(df["Churn"]=="Yes").mean()*100
churn_rate
```

```python
## 2. which contract type has more churn
df.groupby("Contract_Type")["Churn"].value_counts()
```

```python
## to calculate churn percentage by contract type
df.groupby("Contract_Type")["Churn"].apply( lambda x: (x =="Yes").mean()*100)

## Month-to-Month customers has higher churn rate(29.38%) compared with One year(14.76%) and Two year(16.12%) customers
```

```python
## 3.Does Tenure affect Churn

df["Tenure_Group"]=pd.cut(df["Tenure_Months"],bins=[0,12,24,48,72],labels=["0-12","13-24","25-48","49-72"])
df.head(2)
```

```python
## calculate churn by tenure group
df.groupby("Tenure_Group")["Churn"].apply(lambda x:(x=="Yes").mean()*100)

## Churn decreses when customer Tenure increases.Newer customers(0-12 months) have the highest churn rate at 39.87%, while long-term customers(49=72) have the lowest at 14.51%
```

```python
## 4.Do higher monthly charges relate to high churn?

df["Charge_group"]=pd.cut(df["Monthly_Charges"],bins=[0,50,100,150],labels=["Low","Medium","High"])
df.head(2)
```

```python
## calculate churn rate by charge group
df.groupby("Charge_group")["Churn"].apply(lambda x: (x=="Yes").mean()*100)

## Customers with higher monthly charges show higher churn rate.churn increases from 11.53% in low-charge group to 32.64% in high-charge group
```

```python
## 5.Do no of support calls relate to churn?

df["Support_call_group"]=pd.cut(df["Support_Calls"],bins=[-1,1,3,5,10],labels=["0-1","2-3","4-5","6+"])
df["Support_call_group"].value_counts().sort_index()
```

```python
## calculate churn rate by support call group
df.groupby("Support_call_group",observed=True)["Churn"].apply(lambda x:(x=="Yes").mean()*100)

## customers with more support calls have higher churn rate .Churn increases from 18.85% for customers with 0-1 support calls to 33.93% for customers with 6+ calls
```

```python
## 6.Does payment method relate to churn?
df.groupby("Payment_Method")["Churn"].apply(lambda x : (x=="Yes").mean()*100)

## churn varies by payment method .Electronic check has the highest churn rate at 27.83% followed by Bank Transfer at 23.95.Credit card and debit card have lower churn rates at 20.08% and 18.75% respectively.
```

```python
## 7. Does internet service type relate to Churn
df.groupby("Internet_Service")["Churn"].value_counts()
```

```python
## calculate churn rate by internet service
df.groupby("Internet_Service")["Churn"].apply(lambda x:(x=="Yes").mean()*100)

## Churn varies by internet serivice type .Fiber optic customers have the highest churn rate at 26.44% compared with 21.83% with no internet and 18.56%  for DSL customers.
```

```python
## 8.Does age relate to churn?

df["Age_group"]=pd.cut(df["Age"],bins=[17,30,45,60,100],labels=["17-30","31-45","46-60","61+"])
df["Age_group"].value_counts().sort_index()
```

```python
## calculate churn rate by Age group
df.groupby("Age_group")["Churn"].apply(lambda x:(x=="Yes").mean()*100)
```

```python
## 9.Which contract type and internet service combinations has the highest churn 

df.groupby(["Contract_Type","Internet_Service"])["Churn"].value_counts()
```

```python
## calculate churn by combination
df.groupby(["Contract_Type","Internet_Service"])["Churn"].apply(lambda x : (x=="Yes").mean()*100)

## contract type and internet service together  show a clear difference  in churn .Month-to-Month customers have the highest churn across all three internet-Service groups with fiber optic + Month-to-Month  at 32.43%. The lowest rate is DSL +one year at 10.05%.
```

```python
## 10.Does Gender relate to churn?
df.groupby("Gender")["Churn"].value_counts()
```

```python
df.groupby("Gender")["Churn"].apply(lambda x:(x=="Yes").mean()*100)
```

```python
## 11. How do Churned and retained customers differ?

df.groupby("Churn")[["Age","Tenure_Months","Monthly_Charges","Total_Charges","Support_Calls"]].mean()

## Tenure---Churned customers have lower average tenure(28.88% vs 39.10%)
## Monthly Charges---Churned customers pay more on average (89.79% vs 76.81%)
## Support Calls---Churned customers makes more support calls(2.50% vs 2.13%)
## Age---Churned customers are slightly older on average(45.21% vs 43.48%)
## Total Charges---Churned customers have lower charges(2601.41 vs 3028.97),which is consistent with their lower average tenure.
```

```python
## 12. Overall Churn Profile 

categorical_column=["Gender","Contract_Type","Internet_Service","Payment_Method"]

for col in categorical_column:
    print("\n",col)
    print( df.groupby(col)["Churn"].apply( lambda x: (x=="Yes").mean()*100))
```
